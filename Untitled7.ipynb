```python
import requests
import sqlite3

url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)

print(response.status_code)

if response.status_code == 200:
    data = response.json()

    con = sqlite3.connect("store.db")
    cursor = con.cursor()

    for item in data[:5]:
        cursor.execute(
            "INSERT OR IGNORE INTO products (id, name, category, price) VALUES (?, ?, ?, ?)",
            (item["id"], item["title"], "User " + str(item["userId"]), item["id"] * 100)
        )

    con.commit()

    cursor.execute("SELECT * FROM products WHERE category = 'User 1'")
    print(cursor.fetchall())

    cursor.execute("SELECT * FROM products ORDER BY price DESC")
    print(cursor.fetchall())

    con.close()
```
